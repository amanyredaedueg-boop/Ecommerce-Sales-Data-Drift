```python
df['date'] = pd.to_datetime(df['date'])
```

```python
df = df.sort_values('date')
df.head()
```

```python
start_date = df['date'].min()
train_end = start_date + pd.Timedelta(days=20)
train = df[df['date'] <= train_end].copy()
test = df[df['date'] > train_end].copy()

print("Training:", train.shape)
print("Testing:", test.shape)
print("Training period:", train['date'].min(), "to", train['date'].max())
print("Testing period:", test['date'].min(), "to", test['date'].max())
```

```python
features = [
    'price',
    'discount_pct',
    'is_holiday_season'
]
target = 'units_sold'
```

```python
from sklearn.linear_model import LinearRegression
model = LinearRegression()
X_train = train[features]
y_train = train[target]
X_test = test[features]
y_test = test[target]
model.fit(X_train, y_train)
predictions = model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))
r2 = r2_score(y_test, predictions)
print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
test['prediction'] = predictions
test['error'] = abs(test['units_sold'] - test['prediction'])
```

```python
test['week'] = test['date'].dt.to_period('W')
```

```python
weekly_mae = test.groupby('week')['error'].mean()
print(weekly_mae)
```

```python
import matplotlib.pyplot as plt
weekly_mae.plot(kind='line', marker='o')
plt.xlabel('Week')
plt.ylabel('MAE')
plt.title('Weekly MAE - Model Performance')
plt.xticks(rotation=45)
plt.grid()
plt.show()
```

```python
weekly_mae_df = weekly_mae.reset_index()
weekly_mae_df.columns = ['week', 'MAE']
print(weekly_mae_df)
```

```python
weekly_mae_df['MAE_change'] = weekly_mae_df['MAE'].diff()
drift_week = weekly_mae_df.loc[
    weekly_mae_df['MAE_change'].idxmax(), 'week'
]
print("Drift started around:", drift_week)
```

```python
before_drift = test[test['week'] < drift_week]
after_drift = test[test['week'] >= drift_week]
print("MAE Before Drift:",
      before_drift['error'].mean())
print("MAE After Drift:",
      after_drift['error'].mean())
```

```python
print("Average sales before drift:",
      before_drift['units_sold'].mean())
print("Average sales after drift:",
      after_drift['units_sold'].mean())
```

```python
print("Average price before drift:",
      before_drift['price'].mean())
print("Average price after drift:",
      after_drift['price'].mean())
```

```python
print("Average discount before drift:",
      before_drift['discount_pct'].mean())
print("Average discount after drift:",
      after_drift['discount_pct'].mean())
```
